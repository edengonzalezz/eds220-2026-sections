# Water conflicts in the Colorado River Basin


## 1. Archive exploration



a. 

i. Data was collected from public, recorded meetings of water control boards and official organizations in the Colorado River Basin and articles from newspapers describing water-related events in geographic areas in the Basin. 

ii. Observations were collected from 2005-2021.

iii. They believe that this research adds a human dimension for the USGS' Water Mission Area is as it may provide a source of background information to the water availability assessment reports that the USGS produces. 


This qualitative data helps in the understanding of social vulnerabilities and resiliency in the contexxt of water management.


Holloman, D.V., Hines, M.K., and Zoanni, D.K., 2023, Coded Water Conflict and Crisis Events in the Colorado River Basin, Derived from LexisNexis search 2005-2021: U.S. Geological Survey data release, acessed 7 October 2026, [https://doi.org/10.5066/P9X6WR7J](https://doi.org/10.5066/P9X6WR7J).

## Exploratory wrangling

```python
conflict = df.loc[df['Conflict Present'] == 'Y', 'State'].str.split(';').explode()
```

```python
conflict
```

```python
conflict.unique()
```

```python
conflict = conflict.str.strip()
```

```python
conflict_counts = conflict.value_counts(dropna = False)
```

## Visualize data exploration

```python
conflict_counts.plot(kind = 'bar', x = 'State', y = "Conflict Count", title = "Number of Conflicts per State")
```
